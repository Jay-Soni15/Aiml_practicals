```python
import heapq
import time

# Graph with distances
graph = {
    'A': {'B': 2, 'C': 4},
    'B': {'D': 3, 'E': 1},
    'C': {'F': 2},
    'D': {'G': 2},
    'E': {'G': 5},
    'F': {'G': 3},
    'G': {}
}

# Heuristic values
heuristic = {
    'A': 6,
    'B': 4,
    'C': 4,
    'D': 2,
    'E': 2,
    'F': 2,
    'G': 0
}


# Greedy Best First Search
def greedy_search(start, goal):

    pq = [(heuristic[start], start, [start])]
    visited = set()
    nodes_explored = 0

    while pq:

        h, node, path = heapq.heappop(pq)

        if node in visited:
            continue

        visited.add(node)
        nodes_explored += 1

        if node == goal:
            return path, nodes_explored

        for neighbour in graph[node]:

            if neighbour not in visited:
                heapq.heappush(
                    pq,
                    (heuristic[neighbour], neighbour,
                     path + [neighbour])
                )

    return None, nodes_explored


# A* Search
def a_star(start, goal):

    pq = [(heuristic[start], 0, start, [start])]
    best_cost = {start: 0}
    nodes_explored = 0

    while pq:

        f, g, node, path = heapq.heappop(pq)

        if g != best_cost.get(node):
            continue

        nodes_explored += 1

        if node == goal:
            return path, g, nodes_explored

        for neighbour, cost in graph[node].items():

            new_g = g + cost

            if new_g < best_cost.get(neighbour, float('inf')):

                best_cost[neighbour] = new_g
                new_f = new_g + heuristic[neighbour]

                heapq.heappush(
                    pq,
                    (new_f, new_g, neighbour,
                     path + [neighbour])
                )

    return None, float('inf'), nodes_explored


# Starting and Goal Nodes
start = 'A'
goal = 'G'


# Greedy Search Execution
t1 = time.time()

greedy_path, greedy_nodes = greedy_search(start, goal)

greedy_time = time.time() - t1


# A* Execution
t2 = time.time()

astar_path, astar_cost, astar_nodes = a_star(start, goal)

astar_time = time.time() - t2


# Display Results
print("GREEDY BEST FIRST SEARCH")

print("Path:", greedy_path)
print("Nodes Explored:", greedy_nodes)
print("Execution Time:", greedy_time)


print("\nA* SEARCH")

print("Path:", astar_path)
print("Path Cost:", astar_cost)
print("Nodes Explored:", astar_nodes)
print("Execution Time:", astar_time)
```
