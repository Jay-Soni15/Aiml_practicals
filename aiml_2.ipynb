```python
from collections import deque
import time

# Graph representation
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}

# Breadth First Search
def bfs(graph, start, goal):

    queue = deque([[start]])
    visited = set()
    nodes_explored = 0

    while queue:

        path = queue.popleft()
        node = path[-1]

        if node not in visited:

            visited.add(node)
            nodes_explored += 1

            if node == goal:
                return path, nodes_explored

            for neighbour in graph[node]:

                new_path = list(path)
                new_path.append(neighbour)
                queue.append(new_path)

    return None, nodes_explored


# Depth First Search
def dfs(graph, start, goal):

    stack = [[start]]
    visited = set()
    nodes_explored = 0

    while stack:

        path = stack.pop()
        node = path[-1]

        if node not in visited:

            visited.add(node)
            nodes_explored += 1

            if node == goal:
                return path, nodes_explored

            for neighbour in reversed(graph[node]):

                new_path = list(path)
                new_path.append(neighbour)
                stack.append(new_path)

    return None, nodes_explored


# Starting and Goal nodes
start = 'A'
goal = 'G'


# BFS Execution
start_time = time.time()

bfs_path, bfs_nodes = bfs(graph, start, goal)

bfs_time = time.time() - start_time


# DFS Execution
start_time = time.time()

dfs_path, dfs_nodes = dfs(graph, start, goal)

dfs_time = time.time() - start_time


# Display Results
print("BFS Results")
print("Path:", bfs_path)
print("Path Length:", len(bfs_path) - 1)
print("Nodes Explored:", bfs_nodes)
print("Execution Time:", bfs_time)


print("\nDFS Results")
print("Path:", dfs_path)
print("Path Length:", len(dfs_path) - 1)
print("Nodes Explored:", dfs_nodes)
print("Execution Time:", dfs_time)
```
