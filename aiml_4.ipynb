```python
# Knowledge Representation and Expert System

# Knowledge Base: Rules
rules = [
    (["fever", "cough"], "respiratory_infection"),
    (["fever", "headache"], "possible_viral_illness"),
    (["sneezing", "runny_nose"], "possible_allergy"),
    (["fever"], "fever_condition")
]


# Forward Chaining
def forward_chaining(facts):

    print("\n--- Forward Chaining ---")

    new_facts = set(facts)
    changed = True

    while changed:

        changed = False

        for conditions, conclusion in rules:

            if all(condition in new_facts
                   for condition in conditions):

                if conclusion not in new_facts:

                    new_facts.add(conclusion)
                    changed = True

                    print("Rule Applied:", conditions,
                          "->", conclusion)

    return new_facts


# Backward Chaining
def backward_chaining(goal, facts):

    print("\n--- Backward Chaining ---")

    if goal in facts:
        return True

    for conditions, conclusion in rules:

        if conclusion == goal:

            print("Checking Rule:", conditions,
                  "->", conclusion)

            if all(backward_chaining(condition, facts)
                   for condition in conditions):

                return True

    return False


# Main Program
print("MEDICAL EXPERT SYSTEM")

print("\nEnter your symptoms (yes/no):")

symptoms = ["fever", "cough", "headache",
            "sneezing", "runny_nose"]

facts = set()

for symptom in symptoms:

    answer = input("Do you have " + symptom + "? ")

    if answer.lower() == "yes":
        facts.add(symptom)


# Forward Chaining Execution
result = forward_chaining(facts)

print("\nConclusions using Forward Chaining:")

for item in result:
    if item not in facts:
        print(item)


# Backward Chaining Execution
goal = "respiratory_infection"

print("\nGoal to prove:", goal)

if backward_chaining(goal, facts):
    print("Backward Chaining: Goal is TRUE")
else:
    print("Backward Chaining: Goal is FALSE")
```
